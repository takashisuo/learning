### フィルタリング

```python
# データを定義する
df = pd.DataFrame(data = {
    'ID': [1, 2, 3],
    '日本語': ['りんご', 'ぶどう', 'レモン'],
    '英語': ['apple', 'grape', 'lemon']
})
display(df)
print("------------------")

# 「ID」列が2以上のデータを抽出する
filter = df[(df['日本語'] == "オレンジ")]
print(filter)
```

### 集計

```python
# データを定義する
df = pd.DataFrame(data = {
    'ID': [1, 2, 3],
    '日本語': ['りんご', 'ぶどう', 'レモン'],
    '英語': ['apple', 'grape', 'lemon'],
    '値段': [100, 200, 600]
})

# 「値段」列の平均値を計算する
mean = df['値段'].mean()
print(mean)

max = df['値段'].max()
print(max)
```

### 変換

```python

# データを定義する
df = pd.DataFrame(data = {
    'ID': [1, 2, 3],
    '日本語': ['りんご', 'ぶどう', 'レモン'],
    '英語': ['apple', 'grape', 'lemon'],
    '値段': [100, 200, 600]
})

# これくらいならこちらでも可
print(df["値段"] * 10)
df["値段10倍"] = df["値段"] * 10
display(df)

# 値を10倍にしてcostに返す関数「transform」を定義
def transform(cost):
    return cost * 10

# 関数transformを適用し、transform_resultに格納
transform_result= df['値段'].apply(transform)
print(transform_result)
```

## 8章: データ結合

```python
import pandas as pd


# CSVファイルを読み込む
df = pd.read_csv('Chapter7_1.csv')

# [1:3]: 2行目から3行目、[0, 2]: 1列目と3列目を抽出してdf2に格納
df2 = df.iloc[1:3, [0, 2]]

print(df2)
```

### 型変換

```python
import pandas as pd

# データを定義する
df = pd.DataFrame({
    'ID': ['1', '2', '3'],
    'point': ['80', '90', '70']
})

print(df.dtypes)

print("\n---------------------\n")

# point列を数値型に変換する
df['point'] = pd.to_numeric(df['point'])

# df.dtypes: データ型を出力
print(df.dtypes)
```

### 文字列への変換（あまり使わない）

```python
import pandas as pd


# データフレームの作成
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, 28, 22, 35],
    'City': ['New York', 'Los Angeles', 'London', 'Tokyo']
}
df = pd.DataFrame(data)

# to_string()使用前
print("to_string()使用前:")
print(df)
print(type(df))

print("\n---------------------\n")

# to_string()使用後
df_string = df.to_string()
print("to_string()使用後:")
print(df_string)
print(type(df_string))
```

### locの扱い方

```python
import pandas as pd
# データを定義する
df = pd.DataFrame({
    'ID': [1, 2, 3],
    '日本語': ['りんご', 'ぶどう', 'レモン'],
    '英語': ['apple', 'grape', 'lemon']
})
# 列名を指定
name_col = df.loc[:, '日本語']
print(name_col)
print(type(name_col))
```

### 結合

```python
import pandas as pd


# データフレームの作成
df1 = pd.DataFrame({'ID': [1, 2],
                    '日本語': ['りんご', 'ぶどう'],
                    '英語': ['apple', 'grape']
})
display(df1)
print("-----------------------------------")


df2 = pd.DataFrame({'ID': [3, 4],
                    '日本語': ['レモン', 'バナナ'],
                    '英語': ['lemon', 'banana']
})
display(df2)

print("-----------------------------------")

# 縦結合
result = pd.concat([df1, df2], axis=0)

# 結果の表示
display(result)
```
