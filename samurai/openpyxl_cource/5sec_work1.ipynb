```python
import openpyxl
from openpyxl.styles import Border, Side

# 新しいワークブックを作成
workbook = openpyxl.Workbook()
sheet = workbook.active
```

```python
# 表のヘッダー行を書き込む
header = ['商品名', '単価', '数量', '売上']
sheet.append(header)

# 表のデータを書き込む
data = [
    ['商品A', 1000, 8, 8000],
    ['商品B', 2000, 5, 10000],
    ['商品C', 500, 13, 6500]
]
# for文でdataの各行をワークシートに追加する
for row in data:
    sheet.append(row)
    print(row)
```

```python
# セルA1:D4に上下左右に枠線を追加
"""
スタイル名	見た目の説明
thin	標準的な細い実線（デフォルトでよく使われる）
thick	太い実線
medium	中太の実線
dashed	破線
dotted	点線
double	二重線
hair	極細線（thinよりもさらに細い）
"""
thin_border = Border(left=Side(style='thin'),
                    right=Side(style='dashed'),
                    top=Side(style='dashed'),
                    bottom=Side(style='thin'))
# セル範囲をsheet['A1:D4'] と指定して、for文で各セルに枠線を追加
# まずはrow: A1, B1, C1, D1
for row in sheet['A1:D4']:
    print(row)
    # その後、セルごとの加工
    for cell in row:
        cell.border = thin_border

# ファイルを保存
# 左のフォルダのcontent直下に保存される。
# このフォルダはGoogle Colaboratoryが稼働しているOS上となるのでセッションが切れたら消えてしまう
# 必要ならドライブをマウントして保存する
workbook.save("売上表.xlsx")
#workbook.save('/content/drive/Othercomputers/DATACANVAS/github/learning_v2/samurai/openpyxl_cource/売上表.xlsx')
```

### グラフの作成

```python
import openpyxl
from openpyxl.chart import BarChart, Reference

# Excelファイルを読み込む
workbook = openpyxl.load_workbook('売上表.xlsx')
sheet = workbook.active
```

```python
# グラフに使うデータ範囲を指定する（4列目〜4列目、1行目〜最終行目）
data = Reference(sheet, min_row=1, min_col=4, max_row=sheet.max_row, max_col=4)

# グラフに使うデータ範囲を指定する（1列目〜1列目、2行目〜最終行目）
categories_range = Reference(sheet, min_col=1, min_row=2, max_row=sheet.max_row, max_col=1)

# 棒グラフオブジェクトを作成する
chart = BarChart()

# 指定されたデータ範囲をグラフに追加する
chart.add_data(data, titles_from_data=True)

# グラフのX軸に表示するラベルを設定する
chart.set_categories(categories_range)
```

```python
# グラフを挿入する位置を指定（最終行の２つ下の行）
chart_cell = sheet.max_row + 2

# グラフをExcelファイルの指定した位置に挿入
sheet.add_chart(chart, f"A{chart_cell}")

# Excelファイルを保存
workbook.save("売上表_棒グラフ.xlsx")
```
