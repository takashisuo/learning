## データフレームのExcelファイルへの書き込み

```python
# openpyxlの利用時
import pandas as pd
from openpyxl import Workbook
from openpyxl.utils.dataframe import dataframe_to_rows

# データフレームを作成する
df = pd.DataFrame(data = {
    'ID': [1, 2, 3],
    '名前': ['田中一郎', '山田次郎', '加藤三郎'],
    '年齢': [20, 30, 40]
})

# openpyxlでの出力
path = "/content/drive/Othercomputers/DATACANVAS/github/learning_v2/samurai/openpyxl_cource/"
# Excelファイルの作成
wb = Workbook()
ws = wb.active

# データフレームからリストを取得してExcelファイルに追加
for r in dataframe_to_rows(df, index=False, header=True):
    ws.append(r)

wb.save(path + '/' + 'Chapter10_1_openpyxl_output.xlsx')
```

## データのソート

```python
import pandas as pd

# Excelデータをデータフレームに変換
df = pd.read_excel(path + '/' + 'Chapter10_3.xlsx', sheet_name='Sheet1')

# 各教科の合計を算出
df['合計'] = df['国語'] + df['数学'] + df['英語']

# 「合計」列で降順ソート
df = df.sort_values('合計', ascending=False)

display(df)

# データフレームをExcelファイルに書き出し
writer = pd.ExcelWriter('Chapter10_3_openpyxl_output.xlsx')

# データフレームオブジェクトをExcelファイルに書き込む
df.to_excel(writer, sheet_name='Sheet1', index=False)
# Excelファイルを閉じる
writer.close()
```

## データの削除とフィルタリング
